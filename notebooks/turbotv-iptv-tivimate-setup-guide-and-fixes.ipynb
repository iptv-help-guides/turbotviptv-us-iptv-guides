## How do I add my TurboTV IPTV TiviMate playlist via Xtream or M3U?
To get started with your television service, you need to add your playlist credentials into the application. TiviMate offers two primary methods: Xtream Codes and M3U playlists. Xtream Codes is generally faster and organizes video-on-demand content better.

1. Open TiviMate and select 'Add Playlist'.
2. Choose 'Xtream Codes' (recommended) or 'M3U playlist'.
3. For Xtream, enter the Server URL, your Username, and your Password provided by your [IPTV subscription](https://turbotviptv.us/iptv-subscription/). If using M3U, type in the long URL you received.
4. Check the box for 'Include VOD' if you want movies and series.
5. Click 'Next' and allow the app to process the channels.

Typing errors are the leading cause of setup failures. Ensure there are no extra spaces, especially after periods or slashes.

## How can I configure the EPG source and update interval?
The Electronic Program Guide (EPG) is essential for seeing what is currently airing and scheduling future viewings. If you used Xtream Codes, the EPG usually loads automatically. If it is missing, or if you used an M3U link, you must configure it manually.

1. Go to Settings, then select 'EPG'.
2. Choose 'EPG Sources' and click 'Add Source'.
3. Enter the EPG URL provided by your service.
4. Assign this source to your main playlist.

Next, adjust the update interval to ensure your TV guide is always current. In the EPG settings, look for 'Update interval' and set it to update every 12 to 24 hours. Updating too frequently can slow down your device, while updating too rarely leaves you with empty guide data. You can also toggle 'Update on app start' for convenience.

## How do I manage channel groups and set up favorites?
Navigating thousands of channels can be overwhelming, so hiding unwanted categories and creating a favorites list is highly recommended. TiviMate makes group management straightforward.

1. Open your TV guide and long-press the 'Select' or 'OK' button on your remote while highlighting a channel.
2. A menu will appear on the right side of the screen. Select 'Group options' and then 'Manage groups'.
3. Toggle off any international or local categories you do not watch.

To build your favorites list, simply highlight a channel you love in the guide, long-press the 'OK' button, and choose 'Add to Favorites'. A star icon will appear next to the channel name. You can then access all your preferred networks quickly from the dedicated 'Favorites' category at the top of your group list, saving you time every time you launch the app.

## What are the correct catch-up settings for missed broadcasts?
Catch-up allows you to watch previously aired programs without recording them in advance. However, this feature only works if your provider supports it and the application is configured correctly.

1. Navigate to Settings and select 'Playlists'.
2. Choose your active playlist and scroll down to 'Catch-up'.
3. Ensure the 'Catch-up type' is set to 'Xtream Codes' or 'Auto'.
4. Adjust the 'Catch-up duration' to match what your provider offers (typically 1 to 7 days).

When configured properly, you will see a small clock icon next to past programs in the TV guide. Simply click on a past show to start streaming it. If playback buffers or fails to start, verify your internet connection. You need stable speeds of at least 10 Mbps per screen for HD, 25 Mbps for Full HD, and 40+ Mbps for 4K streaming.

## Why is the stream buffering despite a fast internet connection?
Users often experience buffering even when speed tests show high numbers. For live streaming, connection stability matters significantly more than peak speed. Your internet might occasionally drop packets or experience high latency, which disrupts the continuous data feed required for live video.

First, ensure your device is connected via an Ethernet cable rather than Wi-Fi. Wireless connections are susceptible to interference from other household electronics and physical barriers like walls. Second, restart your router and streaming device to clear their internal memory.

If you use TurboTV, you can reach out through their [contact page](https://turbotviptv.us/contact/) if the issue persists across all channels. However, if only one specific channel is buffering while others work flawlessly, the issue is likely on the broadcaster's end. Wait a few hours or try a different channel in the same category.

## How do I fix playlist processing errors during setup?
A 'Failed to process playlist' error is the most frequent roadblock during initial setup. This almost always indicates a typo in your server URL, username, or password. Even a single misplaced capital letter or a hidden space at the end of the URL will block access.

1. Delete the failed playlist attempt from the app.
2. Verify your credentials from your provider's welcome email.
3. Re-type the Server URL carefully, ensuring you include 'http://' or 'https://' exactly as provided.
4. Check that your subscription is active. You can review your account status on the [official site](https://turbotviptv.us/turbotv-iptv-official-site/).
5. Ensure your device is connected to the internet before hitting the next button.

If you are typing a long M3U link, consider using a URL shortener on your phone or computer, or use the Companion app to copy and paste the link directly from a mobile device to avoid remote-control typing errors.

## 🗓️ IPTV EPG (XMLTV) Guide Checker
Reads an XMLTV guide (.xml or .xml.gz), shows coverage per channel, gaps, time-zone offsets, and which playlist channels have guide data.

**How to use it**
1. Keep **USE_DEMO** ticked for an example, or paste your XMLTV guide URL (.xml or .xml.gz).
2. Optional: your M3U playlist URL, to see which channels have no guide data.
3. **Runtime → Run all**.

> 🔒 The guide is read in your Colab session only.

```python
#@title Sources { display-mode: "form" }
USE_DEMO = True  #@param {type:"boolean"}
EPG_URL = ""  #@param {type:"string"}
M3U_URL = ""  #@param {type:"string"}

import gzip, io, re, requests
import xml.etree.ElementTree as ET
from datetime import datetime, timedelta, timezone

def demo_epg():
    start = datetime.now(timezone.utc).replace(minute=0, second=0, microsecond=0)
    def stamp(dt, tz="+0000"):
        return dt.strftime("%Y%m%d%H%M%S") + " " + tz
    out = ['<?xml version="1.0" encoding="UTF-8"?>', '<tv generator-info-name="demo">']
    for cid, name in [("news1.example", "Example News"), ("sport1.example", "Example Sport 1"), ("kids1.example", "Example Kids"), ("movies1.example", "Example Movies")]:
        out.append(f'<channel id="{cid}"><display-name>{name}</display-name></channel>')
    for cid, hours, tz, gap_at in [("news1.example", 48, "+0000", None), ("sport1.example", 24, "+0100", 10), ("kids1.example", 6, "+0000", None)]:
        t = start
        for h in range(hours):
            if gap_at is not None and h == gap_at:
                t += timedelta(hours=2)  # a 2-hour hole in the schedule
            out.append(f'<programme start="{stamp(t, tz)}" stop="{stamp(t + timedelta(hours=1), tz)}" channel="{cid}"><title>Programme {h + 1}</title></programme>')
            t += timedelta(hours=1)
    out.append("</tv>")
    return "\n".join(out).encode()

def fetch(url):
    r = requests.get(url, timeout=60, headers={"User-Agent": "Mozilla/5.0 (EPG checker notebook)"})
    r.raise_for_status()
    data = r.content
    return gzip.decompress(data) if data[:2] == b"\x1f\x8b" else data

raw = demo_epg() if USE_DEMO else fetch(EPG_URL.strip())
print(f"Guide loaded: {len(raw) / 1e6:.2f} MB")
```

```python
#@title Read the guide
import pandas as pd

def parse_time(s):
    s = s.strip()
    dt = datetime.strptime(s[:14], "%Y%m%d%H%M%S")
    m = re.search(r"([+-])(\d{2})(\d{2})$", s)
    offset = "+0000"
    if m:
        offset = m.group(0)
        delta = timedelta(hours=int(m.group(2)), minutes=int(m.group(3)))
        dt = dt - delta if m.group(1) == "+" else dt + delta
    return dt.replace(tzinfo=timezone.utc), offset

channels, progs = {}, []
for event, el in ET.iterparse(io.BytesIO(raw), events=("end",)):
    if el.tag == "channel":
        channels[el.get("id")] = (el.findtext("display-name") or "").strip()
        el.clear()
    elif el.tag == "programme":
        try:
            start, off = parse_time(el.get("start", ""))
            stop, _ = parse_time(el.get("stop", "")) if el.get("stop") else (None, None)
            progs.append((el.get("channel"), start, stop, off))
        except ValueError:
            pass
        el.clear()

p = pd.DataFrame(progs, columns=["channel", "start", "stop", "offset"]).sort_values(["channel", "start"])
p["gap_h"] = (p.groupby("channel")["start"].shift(-1) - p["stop"]).dt.total_seconds() / 3600
per = p.groupby("channel").agg(programmes=("start", "size"), first=("start", "min"), last=("stop", "max"), gaps=("gap_h", lambda g: int((g > 0.25).sum())))
per["hours_covered"] = ((per["last"] - per["first"]).dt.total_seconds() / 3600).round(1)
per.insert(0, "name", [channels.get(c, "") for c in per.index])
now = datetime.now(timezone.utc)
per["days_ahead"] = ((per["last"] - now).dt.total_seconds() / 86400).round(1)
empty = sorted(set(channels) - set(per.index))

print(f"Channels declared: {len(channels)} · with programmes: {len(per)} · programmes: {len(p):,}")
print("Time-zone offsets used:", {k: int(n) for k, n in p["offset"].value_counts().items()})
if empty:
    print("Channels with no programmes:", ", ".join(empty))
display(per.sort_values("days_ahead"))
```

```python
#@title Days of schedule ahead (chart)
import matplotlib.pyplot as plt
top = per.sort_values("days_ahead").head(25)
ax = top["days_ahead"].plot.barh(figsize=(9, max(3, 0.35 * len(top))), color=["#f87171" if d < 1 else "#b6f23a" for d in top["days_ahead"]])
ax.set_yticklabels([f"{n or c}" for c, n in zip(top.index, top["name"])])
ax.axvline(1, color="#94a3b8", linestyle="--")
ax.set_xlabel("Days covered from now")
ax.set_title("Channels with the least schedule ahead (red = under 24 h)")
plt.tight_layout()
plt.show()
```

```python
#@title Compare with the M3U playlist (optional)
if M3U_URL.strip() and not USE_DEMO:
    m3u = requests.get(M3U_URL.strip(), timeout=30).content.decode("utf-8", errors="replace")
    ids = set(re.findall(r'tvg-id="([^"]+)"', m3u))
    missing = sorted(ids - set(per.index))
    print(f"{len(ids)} tvg-id in the playlist · {len(ids) - len(missing)} with guide data · {len(missing)} without")
    if missing:
        print(", ".join(missing[:100]))
else:
    print("Comparison skipped (set M3U_URL and untick USE_DEMO).")
```

## Reading the results
- **Under 24 h ahead** — the guide will be empty tomorrow; the provider needs to refresh the EPG more often.
- **Gaps** — slots with no programme; the player shows “No information”.
- **Several time-zone offsets** — normal when channels come from different countries; if times look shifted, set the time zone in your IPTV app.

## FAQ
**Can I use TiviMate on a Roku or Apple TV?**
No, TiviMate is an Android-exclusive application. It is only available for Android TV, Google TV, Fire TV devices, and Android mobile devices.

**Why does my TV guide say 'No Information'?**
This happens when the EPG source link is missing or expired. You must manually add the EPG URL provided by your service in the app's settings and force an update.

**Is it possible to record live television with this app?**
Yes, the premium version allows you to record live broadcasts. However, you must connect a USB drive or configure a network-attached storage (NAS) device, as most streaming sticks lack sufficient internal storage.
